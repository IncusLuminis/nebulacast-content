# Стирает ли эффективный рост память о зародыше чёрной дыры?

Иллюстрация к статье Sophie Koudmani и др. (2026), *"How to raise a supermassive black hole: interpreting early
JWST AGN with the AESOPICA simulations"*, MNRAS, arXiv:2607.26177 (раздел 3.3 — шкалирующие соотношения и
сравнение фидуциальной и «SE-BoostMax» моделей аккреции).

Главный вопрос статьи: чёрные дыры, которые видит JWST в ранней Вселенной, стартовали сразу тяжёлыми — или
это лёгкие зародыши, которым просто повезло вырасти очень эффективно? Авторы показывают: при скромной
(фидуциальной) модели аккреции разница между лёгкими и тяжёлыми зародышами сохраняется надолго — но стоит
позволить эффективный, саморегулируемый рост (ограниченный лишь потенциалом гало-хозяина), и разница между
стартовыми массами стирается за какие-то сотни миллионов лет.

Здесь мы честно (используя стандартную физику аккреции, а не выдуманные числа) сравниваем два сценария роста
чёрной дыры из четырёх разных зародышей (10², 10³, 10⁴, 10⁵ масс Солнца — тот же диапазон, что использует сама
статья).

**Важная оговорка** (по итогам консультации с Fellow_Astrophysicist, см. `2607.26177_viz_consult_note.md`
в папке статьи): в первой версии этого ноутбука была ошибка — характерное время роста массы чёрной дыры
спутали с «сырым» эддингтоновским временем (~450 млн лет). На самом деле множитель эффективности излучения
ε/(1−ε) сокращает это время примерно до **50 млн лет** при ε=0.1 — иначе рост оказался бы физически абсурдно
быстрым. Ниже используется исправленное значение. Также стоит помнить: реальный рост чёрных дыр в статье —
вспышечный и стохастический, а не гладкая кривая; наш график иллюстрирует итоговый эффект (сходится ли рост
к общему потолку или нет), а не помесячную траекторию из настоящей симуляции.

### Почему «зародыш» для чёрной дыры

Английский термин статьи — **seed** («black hole seed», «seeding channel», «light seed» / «heavy seed»). Мы
перебрали несколько русских вариантов:

- **«Затравка»** — технически самый точный: это устоявшийся термин из материаловедения (метод Чохральского,
  «затравочный кристалл»), обозначающий объект, который вносят извне, чтобы вокруг него начал расти кристалл.
  Ровно та же метафора, что и у английского seed. Проблема — бытовая коннотация: в обычном русском «затравка»
  прочно ассоциируется с «травлей» (буллингом), что режет слух в научном тексте.
- **«Зерно»** — рассматривали, но отклонили: в кристаллографии «зерно» означает нечто другое (отдельный
  кристаллит внутри поликристаллического материала, отделённый «границей зёрен») — использование этого слова
  для seed создало бы путаницу с уже занятым, хорошо знакомым физикам значением.
- **«Зародыш»** — тот вариант, на котором мы в итоге остановились. Это тоже строгий физический термин, просто
  из смежной области — теории нуклеации/фазовых переходов («зародышеобразование», «критический зародыш»), где
  словари прямо дают его синонимом к «центру кристаллизации». Слово может показаться слишком биологическим на
  слух, но по факту оно уже занято точно таким же смыслом в физике кристаллизации.

### Что значит «фидуциальная» модель

В тексте ниже и в самой статье встречается слово **«фидуциальный»** («фидуциальная модель аккреции»,
«фидуциальный сценарий»). Это не выдуманный наукообразный жаргон — слово действительно существует в русском
научном обиходе (от лат. *fiducia*, «доверие»): в статистике есть устоявшийся термин «фидуциальный вывод»
(fiducial inference, метод Фишера). Смысл в физике простой: **фидуциальная модель — это базовый, эталонный
вариант**, который берут за точку отсчёта и с которым потом сравнивают все более смелые/оптимистичные
альтернативы. В статье 2607.26177 «фидуциальная модель аккреции» — это самый скромный, консервативный сценарий
питания чёрной дыры (без сверхэддингтоновских вспышек), в противовес более агрессивным «SE-Boost» и
«SE-BoostMax».

Поскольку слово всё же нечастое и специфическое, ниже по тексту мы для ясности в основном говорим «базовая»
или «эталонная» модель, но хотели явно объяснить, откуда взялось само слово «фидуциальная» — раз уж статья
использует именно его.

```python
from pathlib import Path

import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

OUT_DIR = Path("figures")
OUT_DIR.mkdir(exist_ok=True)

THEME = "dark"  # "light" or "dark" -- controls all plots

if THEME == "dark":
    plt.style.use("dark_background")
elif THEME == "light":
    plt.style.use("default")
else:
    raise ValueError(f"THEME must be 'light' or 'dark', got {THEME!r}")

# --- Eddington / Salpeter growth timescale, standard textbook physics ---
sigma_T = 6.6524587e-25  # Thomson cross-section, cm^2
c_light = 2.99792458e10  # cm/s
G_cgs = 6.674e-8         # cgs
m_p = 1.6726219e-24      # proton mass, g
yr = 3.15576e7           # s

eps = 0.1  # radiative efficiency, the value the paper itself adopts

t_Edd_s = sigma_T * c_light / (4 * np.pi * G_cgs * m_p)
t_Edd_Myr = t_Edd_s / yr / 1e6
tau_Myr = t_Edd_Myr * eps / (1 - eps)  # <-- the correction from the astrophysicist consultation

print(f"'Сырое' эддингтоновское время: {t_Edd_Myr:.0f} млн лет")
print(f"Реальное характерное время роста массы (с учётом eps/(1-eps) при eps={eps}): {tau_Myr:.0f} млн лет")
```

## Два сценария роста

**Сценарий 1 — чистая экспонента** (аналог «фидуциальной», менее эффективной модели аккреции статьи):
масса растёт без ограничения сверху, $dM/dt = M/\tau$. У такого роста нет "потолка" — отношение масс двух
разных зародышей остаётся ровно тем же самым навсегда, сколько бы времени ни прошло.

**Сценарий 2 — саморегулируемый (логистический) рост** (аналог «SE-BoostMax», эффективной модели статьи):
$dM/dt = (M/\tau)\,(1-M/M_{\rm eq})$, где $M_{\rm eq}$ — равновесная масса, заданная гравитационным
потенциалом самого гало-хозяина (та же логика, что и в соотношении M–σ: обратная связь от аккреции
самоограничивает рост чёрной дыры). Это saturating рост: чем ближе чёрная дыра подбирается к $M_{\rm eq}$,
тем медленнее растёт — и именно поэтому разные зародыши в итоге сходятся к одному и тому же потолку,
просто в разное время.

```python
seeds = [1e2, 1e3, 1e4, 1e5]  # Msun -- same range the paper explores
colors = ["#1f77b4", "#2ca02c", "#ff7f0e", "#d62728"]
M_eq = 2e6  # Msun -- illustrative equilibrium mass set by the host halo potential

def rhs_exponential(t, M, tau):
    return [M[0] / tau]

def rhs_logistic(t, M, tau, Meq):
    return [(M[0] / tau) * (1 - M[0] / Meq)]

t_span_Myr = np.linspace(0, 600, 400)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 5), sharey=True)

for M0, col in zip(seeds, colors):
    sol_exp = solve_ivp(rhs_exponential, [0, 600], [M0], t_eval=t_span_Myr, args=(tau_Myr,), rtol=1e-8)
    sol_log = solve_ivp(rhs_logistic, [0, 600], [M0], t_eval=t_span_Myr, args=(tau_Myr, M_eq), rtol=1e-8, atol=1.0)
    ax1.plot(t_span_Myr, sol_exp.y[0], color=col, lw=2.2, label=f"{M0:.0e} $M_\\odot$")
    ax2.plot(t_span_Myr, sol_log.y[0], color=col, lw=2.2, label=f"{M0:.0e} $M_\\odot$")

for ax, title in zip((ax1, ax2), ("Чистая экспонента\n(память о зародыше НЕ стирается)",
                                   "Саморегулируемый рост\n(память о зародыше стирается)")):
    ax.set_yscale("log")
    ax.set_xlabel("Время [млн лет]")
    ax.set_title(title, fontsize=11)
    ax.grid(alpha=0.3, which="both")
ax1.set_ylabel(r"Масса чёрной дыры [$M_\odot$]")
ax2.axhline(M_eq, color="gray", ls=":", lw=1.2)
ax2.text(420, M_eq * 1.3, "$M_{eq}$ (потолок,\nзадан гало-хозяином)", fontsize=8, color="gray")
ax1.legend(title="Зародыш", fontsize=9)
fig.suptitle("Рост чёрной дыры из четырёх разных зародышей: два сценария аккреции")
fig.tight_layout()
fig.savefig(OUT_DIR / "seed_growth_comparison.png", dpi=140)
plt.show()
```

## Что видно на графике

Слева — если рост ничем не ограничен, кривые для разных зародышей идут строго параллельно (в логарифмическом
масштабе — это буквально означает, что расстояние между ними, то есть отношение масс, никогда не меняется).
Зародыш, стартовавший в тысячу раз легче другого, так и останется в тысячу раз легче — сколько бы времени
ни прошло. Это судьба чёрных дыр в менее эффективной («фидуциальной») модели аккреции статьи.

Справа — как только рост начинает саморегулироваться обратной связью (упираясь в потолок, заданный
гравитационным потенциалом самого гало-хозяина, а не начальной массой зародыша), картина меняется
принципиально: все четыре кривые, стартовавшие на четыре порядка величины друг от друга, сходятся
к одному и тому же итоговому значению. Самый тяжёлый зародыш добирается до потолка почти сразу,
самая лёгкая — позже, но все они там оказываются. Заметьте: разница между сценариями — это не разница
в скорости роста как таковой, а разница в том, есть ли у роста предел вообще.

```python
# --- how long does each seed take to reach 90% of the equilibrium mass? ---
target = 0.9 * M_eq
print(f"Время достижения 90% от потолка ({M_eq:.0e} Msun) для разных зародышей:\n")
for M0 in seeds:
    sol = solve_ivp(rhs_logistic, [0, 2000], [M0], args=(tau_Myr, M_eq), dense_output=True, rtol=1e-9, atol=1.0)
    t_fine = np.linspace(0, 2000, 20000)
    M_fine = sol.sol(t_fine)[0]
    idx = np.searchsorted(M_fine, target)
    t90 = t_fine[idx] if idx < len(t_fine) else np.nan
    print(f"  зародыш {M0:.0e} Msun -> {t90:.0f} млн лет")
```

## Итог

Именно этот эффект — способность эффективной, саморегулируемой аккреции «стереть» стартовую массу зародыша
за какие-то сотни миллионов лет — и есть главная методическая трудность статьи 2607.26177: если чёрные дыры
действительно растут настолько эффективно, то одна лишь популяция уже выросших сверхмассивных чёрных дыр,
которую видит JWST, *в принципе не может* сказать нам, с чего они начинали — лёгкие зародыши от звёзд
населения III или сразу тяжёлые зародыши прямого коллапса дают на выходе одну и ту же картину. Именно поэтому
авторы ищут различающие признаки не в самой популяции, а на её низкомассовом краю и в химическом составе
газа галактик-хозяев — там, где эта саморегуляция ещё не успела стереть все следы.
